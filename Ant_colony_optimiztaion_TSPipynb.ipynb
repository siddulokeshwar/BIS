```python

import random
import math

cities = [(0, 0), (1, 5), (5, 2), (6, 6), (8, 3)]
n = len(cities)

# Initialize parameters
ants = 5
iterations = 10
alpha = 1
beta = 2
rho = 0.5
Q = 100

# Calculate distance between cities
distance = [[0] * n for i in range(n)]

for i in range(n):
    for j in range(n):
        distance[i][j] = math.dist(cities[i], cities[j])

# Initialize pheromone
pheromone = [[0.1] * n for i in range(n)]

best_route = None
best_distance = float('inf')

print("Cities:", cities)

# ACO algorithm
for iteration in range(iterations):

    routes = []

    # Construct solutions
    for ant in range(ants):

        start = random.randint(0, n - 1)
        route = [start]
        unvisited = list(range(n))
        unvisited.remove(start)

        while unvisited:

            current = route[-1]

            # Calculate probabilities
            weights = [
                (pheromone[current][j] ** alpha) *
                ((1 / distance[current][j]) ** beta)
                for j in unvisited
            ]

            total = sum(weights)
            probabilities = [w / total for w in weights]

            next_city = random.choices(
                unvisited, weights=probabilities
            )[0]

            route.append(next_city)
            unvisited.remove(next_city)

        # Return to starting city
        route.append(start)

        # Calculate total distance
        total_distance = sum(
            distance[route[i]][route[i + 1]]
            for i in range(n)
        )

        routes.append((route, total_distance))

        # Update best solution
        if total_distance < best_distance:
            best_distance = total_distance
            best_route = route.copy()

    # Update pheromones
    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - rho)

    for route, dist in routes:
        for i in range(n):
            a = route[i]
            b = route[i + 1]
            pheromone[a][b] += Q / dist
            pheromone[b][a] += Q / dist

    # Display iteration results
    print("\nIteration:", iteration + 1)
    print("Best Route:", best_route)
    print("Shortest Distance:", round(best_distance, 2))

# Final output
print("\nFinal Best Route:", best_route)
print("Final Shortest Distance:", round(best_distance, 2))
```

```python
import numpy as np
tes
city_coords = np.array([
    [10, 20],
    [30, 80],
    [20, 50],
    [80, 10],
    [90, 40]
])
num_cities = len(city_coords)


dist_matrix = np.zeros((num_cities, num_cities))
for i in range(num_cities):
    for j in range(num_cities):
        if i != j:
            dist_matrix[i, j] = np.linalg.norm(city_coords[i] - city_coords[j])
        else:
            dist_matrix[i, j] = 1e-10  # Avoid division by zero

num_ants = 5
num_iterations = 20
alpha = 1.0       # Pheromone importance
beta = 2.0        # Distance priority (heuristic)
evaporate = 0.5   # Pheromone evaporation rate
Q = 100           # Pheromone constant


pheromones = np.ones((num_cities, num_cities))

# Track history
best_distance = float('inf')
best_path = []

for iteration in range(num_iterations):
    all_paths = []
    all_distances = []

    for ant in range(num_ants):
        current_city = np.random.randint(num_cities)
        visited = [current_city]
        path_distance = 0

        while len(visited) < num_cities:

            unvisited = [c for c in range(num_cities) if c not in visited]


            numerator = (pheromones[current_city, unvisited] ** alpha) * \
                        ((1.0 / dist_matrix[current_city, unvisited]) ** beta)
            probabilities = numerator / np.sum(numerator)


            next_city = np.random.choice(unvisited, p=probabilities)
            path_distance += dist_matrix[current_city, next_city]
            visited.append(next_city)
            current_city = next_city

        path_distance += dist_matrix[visited[-1], visited[0]]
        visited.append(visited[0])

        all_paths.append(visited)
        all_distances.append(path_distance)

        # Update global best
        if path_distance < best_distance:
            best_distance = path_distance
            best_path = visited


    pheromones *= (1 - evaporate)

    # Deposit new pheromones based on completed routes
    for path, dist in zip(all_paths, all_distances):
        for idx in range(num_cities):
            from_c = path[idx]
            to_c = path[idx+1]
            pheromones[from_c, to_c] += Q / dist
            pheromones[to_c, from_c] += Q / dist

print(f"Best Path indices: {best_path}")
print(f"Best Path Distance: {best_distance:.2f}")
```
